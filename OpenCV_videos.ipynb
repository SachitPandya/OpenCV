#### Video Capture

```python
cap=cv2.VideoCapture(r"C:\Users\BAPS\Desktop\Python\Videos\Virat_CoverDrive.mp4")
"""
cv2.VideoCapture() ->captures the video/camera feed and stores them in the form of frames 
arg -> if 0 -> system webcam
          1 -> external connected cmamera
       if path -> video path
"""             
while True:
    ret,frame=cap.read()
    # cap.read() -> reads the captured feed and returns 2 args
    # ret is flag ->if false something has gone wrong with capture
    # frame -> each frames of the video

    if not ret:
        print("Something went wrong")
        break

    cv2.imshow("Video",frame)

    if cv2.waitKey(1) & 0XFF ==ord("q"):
        print("Quitting")
        break
        
# cap.release()
# cap.release() -> empties the data that is stored in the cap i.e. you have to again initalize cap variable
cv2.destroyAllWindows()
```

#### Saving the Video/feed

```python
frame_width=int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height=int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
# From this we get the width and height of the captured video

codec=cv2.VideoWriter_fourcc(*"mp4v")
# This is encryption / Decryption format in which the output video will be stored 
output=cv2.VideoWriter(r"C:\Users\BAPS\Desktop\Python\Videos\output.mp4",codec,60,(frame_width,frame_height))
"""
cv2.VideoWriter(file_name,code,fps,(width,height))
arg1 -> path of output file/camera_feed[0,1,....]
arg2 -> encryption code/compression format
arg3 -> fps rate 
arg4 -> this is part that is extracted from the input video
"""
while True:
    ret,frame=cap.read()

    output.write(frame)
    cv2.imshow("Saved Video",frame)
    if cv2.waitKey(1) == ord('q'):
        break

output.release()
cv2.destroyAllWindows()
```

#### Extracting particular frames from a video 

```python
ret,frame=cap.read()
fps=cap.get(cv2.CAP_PROP_FPS)
# This function returns the FPS of the video
second=19.5

frame_sec=second*fps    # From this we can get the frame at that particular second
total_frames=cap.get(cv2.CAP_PROP_FRAME_COUNT)
# Calculates the total number of frames there are in the video
if frame_sec<total_frames:
    cap.set(cv2.CAP_PROP_POS_FRAMES,frame_sec)
    """
    cap.set(cv2.CAP_PROP_POS_FRAMES,frames) -> sets the starting frame of an video at an particular frame
    arg1 -> cv2.CAP_PROP_POS_FRAMES = constant indicating frame 0
    arg2 -> frame = it tells the set method to jump to this frame
    """
    ret,frame=cap.read()
    if ret:
        cv2.imshow("Frame at the time ",frame)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    else:
        print("Something went wrong with the video")
else:
    print("Time given is longer than video length")
```

#### Object detection using templates

```python
cap=cv2.VideoCapture(0)
template_img=cv2.imread(r"C:\Users\BAPS\Pictures\Camera Roll\WIN_20250929_13_24_26_Pro.jpg")
h,w=template_img.shape[:2]
methods=[cv2.TM_CCOEFF,cv2.TM_CCOEFF_NORMED,cv2.TM_CCORR,cv2.TM_CCORR_NORMED,cv2.TM_SQDIFF,cv2.TM_SQDIFF_NORMED]
method=cv2.TM_CCORR_NORMED
while True:
    ret,frame=cap.read()
    if not ret:
        print("OOPS!!!")
        break
    # for method in methods:
    img2=frame.copy()
    result= cv2.matchTemplate(frame,template_img,method)
    min_val,max_val,min_loc,max_loc=cv2.minMaxLoc(result)

    if method in [cv2.TM_SQDIFF,cv2.TM_SQDIFF_NORMED]:
        location = min_loc
    else:
        location=max_loc
    bottom_right=(location[0]+w,location[1]+h)
    string=f"Match {[method]}"
    cv2.rectangle(frame,location,bottom_right,255,5)
    cv2.imshow(string,img2)
    if cv2.waitKey(1) & 0XFF==ord("q"):
        print("Quitting")
        break

cap.release()
cv2.destroyAllWindows()
```

#### MeanShift()
- Checks the every instance of the video is checked in form of pixel distribution. 
- We define a initial window for which we provide the coordinates,which defines the area of maximum pixel distribution and tries to keep track of that area in an moving video
- Direction deopends on the difference between the center of window and the centroid of all k-points inside the window
    Direction = (Center - centroid)
- It can differentiate between static backgrund and movinf foreground
- It is kind of tracking algorithm

##### Here rather than giving the coordinates manually i will use template matching for tracking of the ball

```python
import cv2
import numpy as np

cap = cv2.VideoCapture(r"C:\Users\BAPS\Desktop\Python\Videos\Passing_video.mp4")
football_template_img = cv2.imread(r"football_template.png")
h, w = football_template_img.shape[:2]
method = cv2.TM_SQDIFF

# Read first frame
ret, frame = cap.read()
if not ret:
    print("Video not found or cannot be opened!")
    exit()

# Initial template matching
result = cv2.matchTemplate(frame, football_template_img, method)
min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(result)
if method in [cv2.TM_SQDIFF, cv2.TM_SQDIFF_NORMED]:
    location = min_loc
else:
    location = max_loc

bottom_right = (location[0] + w, location[1] + h)
roi = frame[location[1]:bottom_right[1], location[0]:bottom_right[0]]

# Histogram for ROI (in HSV space)
hsv_roi = cv2.cvtColor(roi, cv2.COLOR_BGR2HSV)
mask = cv2.inRange(hsv_roi, np.array((0., 60., 32.)), np.array((180., 255., 255.)))
roi_hist = cv2.calcHist([hsv_roi], [0], mask, [180], [0, 180])
cv2.normalize(roi_hist, roi_hist, 0, 255, cv2.NORM_MINMAX)

# Initial tracking window
trackwindow = (location[0], location[1], w, h)

# Termination criteria
term = (cv2.TERM_CRITERIA_EPS | cv2.TERM_CRITERIA_COUNT, 15, 2)

while True:
    ret, frame = cap.read()
    if not ret:
        print("OOPS!!! End of video")
        break

    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
    bp = cv2.calcBackProject([hsv], [0], roi_hist, [0, 180], 1)

    ret, trackwindow = cv2.meanShift(bp, trackwindow, term)

    x, y, w, h = trackwindow
    vid = cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 0, 255), 2)

    cv2.imshow("Ball Tracker", vid)

    if cv2.waitKey(30) & 0xFF == ord("q"):
        print("Quitting")
        break

cap.release()
cv2.destroyAllWindows()
```

#### CamShift()
- Continously Adaptive Meanshift 
- First applies meanshift ,once meanshift converges,it updates the size of window,find the orientation of best fitting ellipse .
- Againg applies the meanshift with new scaled window and previous window location
- This process is continues until the required conditions are met.

#### Color Detection in Videos

```python
import cv2
import numpy as np
```

```python
cap=cv2.VideoCapture(0)

lower_red1 = np.array([0, 120, 70])
upper_red1 = np.array([10, 255, 255])

lower_red2 = np.array([170, 120, 70])
upper_red2 = np.array([180, 255, 255])

while True:
    ret,frame=cap.read()
    if not ret:
        print("OOPS!!!")

    hsv_frame=cv2.cvtColor(frame,cv2.COLOR_BGR2HSV)

    mask1 = cv2.inRange(hsv_frame, lower_red1, upper_red1)  
    mask2 = cv2.inRange(hsv_frame, lower_red2, upper_red2)  
    mask = mask1 | mask2  

    red_img = cv2.bitwise_and(frame, frame, mask=mask)

    cv2.imshow("Red Objects",red_img)
    cv2.imshow("Real Feed",frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        print("Quitting")
        break

cap.release()
cv2.destroyAllWindows()
```
