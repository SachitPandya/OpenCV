#### Video Processing

```python
import cv2 
import numpy as np
```

#### Video Capture

```python
cap=cv2.VideoCapture(r"C:\Users\BAPS\Desktop\Python\Videos\Virat_CoverDrive.mp4")
"""
cv2.VideoCapture() ->captures the video/camera feed and stores them in the form of frames 
arg -> if 0 -> system webcam
          1 -> external connected cmamera
       if path -. video path
"""             
while True:
    ret,frame=cap.read()
    # cap.read() -> reads the captured feed and returns 2 args
    # ret is flag ->if false something has gone wrong with capture
    # frame -> each frames of the video

    if not ret:
        print("Something went wrong")
        break

    cv2.imshow("Video",frame)

    if cv2.waitKey(1) & 0XFF ==ord("q"):
        print("Quitting")
        break
        
# cap.release()
# cap.release() -> empties the data that is stored in the cap i.e. you have to again initalize cap variable
cv2.destroyAllWindows()
```

#### Saving the Video/feed

```python
frame_width=int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height=int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
# From this we get the width and height of the captured video

codec=cv2.VideoWriter_fourcc(*"mp4v")
# This is encryption / Decryption format in which the output video will be stored 
output=cv2.VideoWriter(r"C:\Users\BAPS\Desktop\Python\Videos\output.mp4",codec,60,(frame_width,frame_height))
"""
cv2.VideoWriter(file_name,code,fps,(width,height))
arg1 -> path of output file/camera_feed[0,1,....]
arg2 -> encryption code/compression format
arg3 -> fps rate 
arg4 -> this is part that is extracted from the input video
"""
while True:
    ret,frame=cap.read()

    output.write(frame)
    cv2.imshow("Saved Video",frame)
    if cv2.waitKey(1) == ord('q'):
        break

output.release()
cv2.destroyAllWindows()
```

#### Extracting particular frames from a video 

```python
ret,frame=cap.read()
fps=cap.get(cv2.CAP_PROP_FPS)
# This function returns the FPS of the video
second=19.5

frame_sec=second*fps    # From this we can get the frame at that particular second
total_frames=cap.get(cv2.CAP_PROP_FRAME_COUNT)
# Calculates the total number of frames there are in the video
if frame_sec<total_frames:
    cap.set(cv2.CAP_PROP_POS_FRAMES,frame_sec)
    """
    cap.set(cv2.CAP_PROP_POS_FRAMES,frames) -> sets the starting frame of an video at an particular frame
    arg1 -> cv2.CAP_PROP_POS_FRAMES = constant indicating frame 0
    arg2 -> frame = it tells the set method to jump to this frame
    """
    ret,frame=cap.read()
    if ret:
        cv2.imshow("Frame at the time ",frame)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    else:
        print("Something went wrong with the video")
else:
    print("Time given is longer than video length")
```
