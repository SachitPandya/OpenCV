### Different types of Interpolation 

```python
#INTER_LINEAR -> Used for Zomming or for general purpose ,Bilinear interpolation ,Balances speed and quality
int_lnr=cv2.resize(img,None,fx=0.5,fy=0.7,interpolation=cv2.INTER_LINEAR)
cv2.imshow("Resized Image Linear",int_lnr)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#INTER_AREA -> Used for Shrinking image,pixel area relation,Best for downscaling ,minimizes distortion
int_ar=cv2.resize(img,None,fx=0.5,fy=0.7,interpolation=cv2.INTER_AREA)
cv2.imshow("Resized Image Area",int_ar)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#INTER_CUBIC -> Used for zomming with high quality,Bicubic interpolation (uses 4×4 pixel neighborhood) ,It is slower but better at zooming images
int_cub=cv2.resize(img,None,fx=0.5,fy=0.7,interpolation=cv2.INTER_CUBIC)
cv2.imshow("Resized Image Cubic",int_cub)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#INTER_NEAREST -> Fast Resizing,Nearest-neighbor interpolation, Fast but low quality and can cause blocky results
int_near=cv2.resize(img,None,fx=0.5,fy=0.7,interpolation=cv2.INTER_NEAREST)
cv2.imshow("Resized Image Nearest",int_near)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Rotating Image using Transformation matrix

```python
h,w=img.shape[:2]
M=cv2.getRotationMatrix2D((w//2,h//2),180,1)
"""
cv2.getRotationMatrix2D(center,angle,scale) -> Function is the transformation matrix on which basis the image is rotated
arg1 -> is the center along which the img will be rotated
arg2 -> is the angle by which img will be rotated 
    +ve angle -> img will be rotated counter clockwise
    -ve angle -> img will be rotated clockwise
arg3 -> scaling factor of the image
"""
rotated_img=cv2.warpAffine(img,M,(w,h),borderMode=cv2.BORDER_CONSTANT,borderValue=(255,255,255))
"""cv2.warpAffine(img_name,transformation,size,flags,borderMode,borderValue)
arg1- > namen of img
arg2 -> M = Transformation matrix
arg3 -> size of output img
arg4 -> Interpolation flags
arg5 -> borderMode = fills the color with are outside img with color
arg6 -> tells the borderMode -> cv2.BORDER_CONSTANT which color to be filled
""" 
cv2.imshow("Rotated Image",rotated_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### BorderMode Types

```python
#cv2.BORDER_CONSTANT() -> fills the background with constant color 
rotated_img_brc=cv2.warpAffine(img,M,(w,h),borderMode=cv2.BORDER_CONSTANT,borderValue=(255,255,255)) 
cv2.imshow("Rotated Image using Border Constant",rotated_img_brc)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# cv2.cv2.BORDER_REPLICATE() -> fills the color of the nearest pixel  
rotated_img_brp=cv2.warpAffine(img,M,(w,h),borderMode=cv2.BORDER_REPLICATE) 
cv2.imshow("Rotated Image using Border Replicate",rotated_img_brp)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# cv2.BORDER_REFLECT() -> Mirros the pixels of nearest edge 
rotated_img_ref=cv2.warpAffine(img,M,(w,h),borderMode=cv2.BORDER_REFLECT) 
cv2.imshow("Rotated Image using Border Reflect",rotated_img_ref)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# cv2.BORDER_REFLECT_101()/cv2.BORDER_REFLECT101() -> Works similar to BORDER_REFLECT but ,if doesn't mirror the border pixel
rotated_img_ref_101=cv2.warpAffine(img,M,(2000,2000),borderMode=cv2.BORDER_REFLECT_101) 
cv2.imshow("Rotated Image using Border Reflect 101",rotated_img_ref_101)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# cv2.BORDER_DEFAULT() -> Works similar to BORDER_REFLECT_101
rotated_img_def=cv2.warpAffine(img,M,(w,h),borderMode=cv2.BORDER_DEFAULT) 
cv2.imshow("Rotated Image using Border Reflect 101",rotated_img_def)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### Changing Perspective of Image

```python
pts1 = np.float32([[1200, 600], [640, 600],
                       [0, 400], [640, 400]])
pts2 = np.float32([[10, 100], [400, 0],
                    [0, 640], [400, 500]])
```

```python
mat=cv2.getPerspectiveTransform(pts1,pts2)
"""
cv2.getPerspectiveTransform(mat1,mat2) -> give the Perspective matrix 
arg1 -> The coordinates of the part of the image you want
arg2 -> The coordinates of border pixels of output img
"""
res=cv2.warpPerspective(img,mat,(w,h),borderMode=cv2.BORDER_REFLECT)
"""
cv2.warpPerspective(img_name,matrix,size)
arg1 -> name of the image
arg2 -> Perspective matrix
arg3 -> size of output image
arg4 & arg5 -> BorderMode and BorderValue
"""
cv2.imshow("Changed the Perspective of the img",res)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
hsv_img=cv2.cvtColor(resized_img,cv2.COLOR_BGR2HSV)
low_red=np.array([0,120,70])
high_red=np.array([15,255,255])

mask=cv2.inRange(hsv_img,low_red,high_red)
"""
cv2.inRange(img,low_limit,high_limit) -> creates an binary mask based on the range specified
arg1 -> image
arg2 -> lower limit of the array 
arg3 -> higher limit of the array
"""
red_img=cv2.bitwise_and(resized_img,resized_img,mask=mask)
cv2.imshow("Red Color in Image",red_img)
cv2.imshow("Original Image",resized_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
