#### Drawing of shapes using mouse

```python
import cv2
import numpy as np
```

```python
events = [i for i in dir(cv2) if 'EVENT' in i]
print( events )
# This will create circles when we double tap the on the screen with Lef mouse button
def draw_circles(events,x,y,flag,param):
    if events == cv2.EVENT_LBUTTONDBLCLK:
        cv2.circle(img,(x,y),100,(200,200,110),5)

img=np.zeros((500,500,3),np.uint8)
cv2.namedWindow("Image")
"""
cv2.namedWindow(window_name,flag)
arg1 -> Window name that will display videos or image
arg2 -> window size is auto adjustable or not 
     -> WINDOW_NORMAL - Allows to manually change window size
     -> WINDOW_AUTOSIZE(Default) - Automatically sets the window size
     -> WINDOW_FULLSCREEN - Changes the window size to fullscreen
"""
cv2.setMouseCallback("Image",draw_circles,3)
"""
cv2.setMouseCallback(window_name,onMouse)
arg1 -> Name of the window
arg2 -> callback function for mouse events
"""
while True:
    cv2.imshow("Image",img)

    if cv2.waitKey(1) & 0XFF==ord("q"):
        print("Quitting")
        break
cv2.destroyAllWindows()
```

```python
draw=False
mde=True
ix,iy=-1,-1

def draw_circles(events,x,y,flag,param):
    global ix,iy,draw,mde

    if events==cv2.EVENT_LBUTTONDOWN:
        draw=True
        ix,iy=x,y

    elif events==cv2.EVENT_MOUSEMOVE:
        if draw==True:
            if mde==True:
                cv2.rectangle(img,(ix,iy),(x,y),(200,100,20),-1)
            else:
                cv2.circle(img,(x,y),10,(20,200,100),-1)
    
    elif events==cv2.EVENT_LBUTTONUP:
        drawing = False
        if mde == True:
            cv2.rectangle(img,(ix,iy),(x,y),(0,255,0),-1)
        else:
            cv2.circle(img,(x,y),5,(0,0,255),-1)

img = np.zeros((512,5120,3), np.uint8)
cv2.namedWindow('image',cv2.WINDOW_FULLSCREEN)
cv2.setMouseCallback('image',draw_circles)

while(1):
    cv2.imshow('image',img)
    k = cv2.waitKey(1) & 0xFF
    if k == ord('m'):
        mde = not mde
    elif k == 27:
        break

cv2.destroyAllWindows()
```

#### Trackbar
- Creates and Trackbar slider with specified name and range(values)

```python
import cv2
import numpy as np

# Create black image with 3 channels
img = np.zeros((1000,1000,3), np.uint8)
cv2.namedWindow("image")

def nothing(x):
    pass

# Trackbars for R, G, B
cv2.createTrackbar("R","image",0,255,nothing)
"""
cv2.createTrackbar(trackbar_name,window_name,value,count,onChange) -> creates a trackbar 
arg1 -> name of trackbar
arg2 -> name of the window
arg3 -> integer that will be changed when trackbar is moved
arg4 -> maximum value of the trackbar
arg5 -> function that will be called every time slider changes value
"""
cv2.createTrackbar("G","image",0,255,nothing)
cv2.createTrackbar("B","image",0,255,nothing)
cv2.setTrackbarMin("B","image",1)
"""
cv2.setTrackbarMin(Trackbar_name,Window_name,new_min_value) -> Sets the minimum value for an trackbar
arg1 -> name of trackbar
arg2 -> name of the window
arg3 -> sets the new low value 
"""
# Switch trackbar
switch = '0: OFF \n1: ON'
cv2.createTrackbar(switch,"image",0,1,nothing)

while True:
    r = cv2.getTrackbarPos('R','image')
    """
    cv2.getTrackbarPos(trackbarname,window_name) -> returns the position of the trackbar
    """
    g = cv2.getTrackbarPos('G','image')
    b = cv2.getTrackbarPos('B','image')
    s = cv2.getTrackbarPos(switch,'image')
 
    if s == 0:
        img[:] = 0
    else:
        img[:] = [b, g, r]

    cv2.imshow("image", img)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cv2.destroyAllWindows()
```

### Morphological Transformations
- Simple operations based on image shape , applied on binary image

```python
import cv2
import numpy as np
import os
```

```python
os.getcwd()
```

```python
img=cv2.imread(r"cropped_image_roi.png",0)
cv2.imshow("Image",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### 1. Erosion 
- It removes the boundries of the foreground object
- It works as, it traverses through the image and if all the pixels under the kernek are 1, then only will keep the pixel as white, otherwise turns white into black(eroded)

- Usefull in removing small noise and detaching two connected objects

```python
kernel=np.ones((3,3),np.uint8)
erod_img=cv2.erode( img,kernel,iterations=2)
"""
cv2.erode(image,kernel,anchor,iterations,border_type,border_value)
arg1 -> image
arg2 -> structuring element used for erosion
arg4 -> number of times erosion is applied
arg5 -> BORDER_WRAP is not supported     
"""
cv2.imshow("Image",img)
cv2.imshow("Erosion Image",erod_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
rect_kernel=cv2.getStructuringElement(cv2.MORPH_RECT,(3,3))
"""
cv2.getStructuringElement(shape,size,anchor)
arg1 -> shape of the kernel
arg2 -> size of kernel
"""
erod_img_rk=cv2.erode(img,rect_kernel,iterations=2)
cv2.imshow("Rectangle kernel Eroded Image",erod_img_rk)
cv2.imshow("Image",img)
cv2.imshow("Erosion Image",erod_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

##### Shapes of kernel

```python
# Rectangle Kernel 
rect_kernel=cv2.getStructuringElement(cv2.MORPH_RECT,(3,3))
erod_img_rk=cv2.erode(img,rect_kernel,iterations=2)
cv2.imshow("Rectangle kernel Eroded Image",erod_img_rk)
cv2.imshow("Image",img)
cv2.imshow("Erosion Image",erod_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Ellipse Kernel 
ellipse_kernel=cv2.getStructuringElement(cv2.MORPH_ELLIPSE,(3,3))
erod_img_re=cv2.erode(img,ellipse_kernel,iterations=2)
cv2.imshow("Ellipse kernel Eroded Image",erod_img_re)
cv2.imshow("Image",img)
cv2.imshow("Erosion Image",erod_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
# OBSERVATION -> It keeps the fine details of fore ground better than rect kernel,but less at efficiently eroding the image
```

```python
# Cross Kernel 
cross_kernel=cv2.getStructuringElement(cv2.MORPH_CROSS,(3,3))
erod_img_rc=cv2.erode(img,cross_kernel,iterations=2)
cv2.imshow("Cross kernel Eroded Image",erod_img_rc)
cv2.imshow("Image",img)
cv2.imshow("Erosion Image",erod_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Diamond Kernel 
diamond_kernel=cv2.getStructuringElement(cv2.MORPH_CROSS,(3,3))
erod_img_rd=cv2.erode(img,diamond_kernel,iterations=2)
cv2.imshow("Diamond kernel Eroded Image",erod_img_rd)
cv2.imshow("Erosion Image",erod_img)
cv2.imshow(" Image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# All Four kernels 
cv2.imshow("Diamond kernel Eroded Image",erod_img_rd)
cv2.imshow("Cross kernel Eroded Image",erod_img_rc)
cv2.imshow("Ellipse kernel Eroded Image",erod_img_re)
cv2.imshow("Rectangle kernel Eroded Image",erod_img_rk)
cv2.imshow("Image",img)
cv2.imshow("Erosion Image",erod_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
# OBSERVATION -> Cross Kernel is the clearest 
```

#### 2. Dilate
- If even "1" pixel is under kernel the it transform all the pixels under the kernel area white
- Increase the region of the foreground/ increase the size of the object

- generally erosion is followed by dilate as, erosion removes white noise but also reduces the area of object. usefull in connecting parts of the object

```python
kernel=np.ones((3,3),np.uint8)
dilate_img=cv2.dilate( img,kernel,iterations=2)
"""
cv2.dilate(image,kernel,anchor,iterations,border_type,border_value)
arg1 -> image
arg2 -> structuring element used for dilation
arg4 -> number of times erosion is applied
arg5 -> BORDER_WRAP is not supported     
"""
cv2.imshow("Image",img)
cv2.imshow("Dilate Image",dilate_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### 3. Opening 
- It's just erosion followed by dilate

```python
open_img=cv2.morphologyEx(img,cv2.MORPH_OPEN,kernel)
"""
cv2.morphologyEx(img,operation,kernel,anchor,iteration,border_type,border_value)
arg 2 -> operation =which operation to be used
"""
cv2.imshow("Image",img)
cv2.imshow("Open Image",open_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### 4. Closing
- Dilate followed by erosion,i.e. opposite of Opening

```python
close_img=cv2.morphologyEx(img,cv2.MORPH_CLOSE,kernel)
cv2.imshow("Image",img)
cv2.imshow("Close Image",close_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cv2.imshow("Open Image",open_img)
cv2.imshow("Close Image",close_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
# OBSERVATION -> Dark color texts are better readable in close_img
```

#### 5. Morphological Gradient
- Difference between the erosion and dilation 

```python
morph_grad_img=cv2.morphologyEx(img,cv2.MORPH_GRADIENT,kernel)
cv2.imshow("Gradient Image",morph_grad_img)
cv2.imshow("Image",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### 6. Top Hat
- Difference between input image and Opening of the image

```python
top_img=cv2.morphologyEx(img,cv2.MORPH_TOPHAT,kernel)
cv2.imshow("TOPHAT Image",top_img)
cv2.imshow("Image",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### 7. Black Hat
- Difference between image and closing of the image

```python
black_hat_img=cv2.morphologyEx(img,cv2.MORPH_BLACKHAT,kernel)
cv2.imshow("BLACK HAT Image",black_hat_img)
cv2.imshow("Image",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cv2.imshow("BLACK HAT Image",black_hat_img)
cv2.imshow("TOPHAT Image",top_img)
cv2.imshow("Image",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
