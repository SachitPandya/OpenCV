#### This file has operations related to contours 

#### Contours 
- Curve joining the joining all the continous points on theboundary , having the same color or intensity 
- Outline or edge of shapes
- Usefull in shape analysis and object detection

```python
import cv2
import numpy as np
```

```python
img=np.zeros((400,400,3),np.uint8)
cv2.rectangle(img,(30,150),(350,350),(255,255,255),-1)
img=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
```

```python
canny_edge=cv2.Canny(img,100,220)
cv2.imshow("Image",img)
cv2.imshow("Canny Image",canny_edge)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
contours,hierarchy=cv2.findContours(img,cv2.RETR_LIST,cv2.CHAIN_APPROX_SIMPLE,offset=(2,2))
""""
cv2.findContours(img,retrival_mode,approximation_mode,offset)
arg1 -> image
arg2 -> determines how contours are retrieved
arg3 -> decides how much detail to keep
arg4 -> point by which every contour point is shifted
     -> Usefull when working with ROI and contour is calculated based on whole image
"""
# """
# returns 
# 1. Contours -> a list of contours in (x,y) 
# 2. Hierarchy -> information about which contour is inside which
# """
cv2.imshow('Canny Edges After Contouring', img)
print("Number of Contours Found = " + str(len(contours)))
cont_img=cv2.drawContours(img,contours,-1,200,3)
"""
cv2.drawContours(img,contours,contour_index,color,thickness,line_type,hierarchy,maxlevel,offset)
arg1 -> image
arg2 -> contours 
arg3 -> which contour to draw,if value is (-)ve draws all contours
arg4 -> color of the contour
arg5 -> thickness of contour line
arg6 -> which line_type to use
arg7(optional) -> needed if want to draw specific contours
arg8 -> if 0 draws only the specific contours
     -> if 1 draws all the specified contours and nested contours
     -> if 2 draws all the specified contours and all the contours inside the contours
arg9 -> shift of the contours
"""
cv2.imshow("Drawing of contours",cont_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### Types of Approximation Modes

```python
#CHAIN_APPROX_SIMPLE -> only the boundary points of the contour is stored 
img=np.zeros((400,400,3),np.uint8)
cv2.rectangle(img,(30,150),(350,350),(255,255,255),-1)
canny_edge=cv2.Canny(img,100,220)
cv2.imshow("Image",img)


contours,hierarchy=cv2.findContours(canny_edge,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE,offset=(2,2))
cont_img_simple=cv2.drawContours(img,contours,-1,(255,220,160),3)
cv2.imshow("Drawing of contours using Simple",cont_img_simple)
cv2.imshow('Canny Edges After Contouring', canny_edge)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# CHAIN_APPROX_NONE -> Stores all the edges of the contour
img=np.zeros((400,400,3),np.uint8)
cv2.rectangle(img,(30,150),(350,350),(255,255,255),-1)
canny_edge=cv2.Canny(img,100,220)
cv2.imshow("Image",img)


contours,hierarchy=cv2.findContours(canny_edge,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_NONE,offset=(2,2))
cont_img_none=cv2.drawContours(img,contours,-1,(255,00,00),3)
cv2.imshow("Drawing of contours using None",cont_img_none)
cv2.imshow('Canny Edges After Contouring', canny_edge)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
contours_n,hierarchy=cv2.findContours(canny_edge,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_NONE,offset=(20,20))
cv2.drawContours(img,contours,-1,(255,00,00),3)
cv2.imshow("Drawing of contours using None",img)

contours_s,hierarchy=cv2.findContours(canny_edge,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE,offset=(2,2))
cv2.drawContours(img,contours,-1,(255,00,00),3)
cv2.imshow("Drawing of contours using Simple",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
# OBSERVATION -> Visually there will be no difference between SIMPLE and NONE ,only difference you would see will be how many contours points are stored
```

### Types of Contour Retrival Modes

```python
"""
Array value format of Hierarchy [Next,Prevoius,First_Child,Parent]
Next -> denotes the next counter at the same level, if no next counter at same level put (-1)
Previous -> denotes the pervious counter at the same level,if no previous counter put (-1)
First_child -> denotes the first child contour, if no child put(-1)
Parent -> denotes the index of Parent counter, if no parent put(-1)
"""
```

```python
# RETR_LIST -> It retrives all the contours and doesn't care about relations between contours
img=np.zeros((400,400,3),np.uint8)
cv2.rectangle(img,(30,150),(350,350),(200,100,155),-1)
cv2.circle(img,(200,200),50,(200,100,100),-1)
canny_edge=cv2.Canny(img,100,220)
cv2.imshow("Image",img)


contours,hierarchy=cv2.findContours(canny_edge,cv2.RETR_LIST,cv2.CHAIN_APPROX_SIMPLE,offset=(2,2))
cont_img_retr_lst=cv2.drawContours(img,contours,-1,(255,220,160),5)
cv2.imshow("Drawing of contours using Simple",cont_img_retr_lst)
cv2.imshow('Canny Edges After Contouring', canny_edge)
cv2.waitKey(0)
cv2.destroyAllWindows()
print(hierarchy)
# AS there is no parent-child relation Third and Fourth term of hierarchy will always be -1
```

```python
# RETR_EXTERNAL -> Retrives only the extreme outside contours and not it's nested contours
img=np.zeros((400,400,3),np.uint8)
cv2.rectangle(img,(30,150),(350,350),(200,100,155),-1)
cv2.circle(img,(200,200),50,(200,100,100),-1)
canny_edge=cv2.Canny(img,100,220)
cv2.imshow("Image",img)


contours,hierarchy=cv2.findContours(canny_edge,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE,offset=(2,2))
cont_img_retr_ext=cv2.drawContours(img,contours,-1,(255,220,160),5)
cv2.imshow("Drawing of contours using Simple",cont_img_retr_ext)
cv2.imshow('Canny Edges After Contouring', canny_edge)
cv2.waitKey(0)
cv2.destroyAllWindows()
print(hierarchy)
# Third and Fourth term of hierarchy will always be -1, As only the outermost contours are taken and rest are ignored
```

```python
# RETR_CCOMP -> Retrives all contours and arrange them in 2 level hierarchy, outer is -1,inside that is -2,and contour inside that is then again-1 
img=np.zeros((400,400,3),np.uint8)
cv2.rectangle(img,(30,150),(350,350),(200,100,155),-1)
cv2.circle(img,(220,240),50,(200,100,100),-1)
canny_edge=cv2.Canny(img,100,220)
cv2.imshow("Image",img)


contours,hierarchy=cv2.findContours(canny_edge,cv2.RETR_CCOMP,cv2.CHAIN_APPROX_SIMPLE,offset=(2,2))
cont_img_retr_ext=cv2.drawContours(img,contours,-1,(255,220,160),5)
cv2.imshow("Drawing of contours using Simple",cont_img_retr_ext)
cv2.imshow('Canny Edges After Contouring', canny_edge)
cv2.waitKey(0)
cv2.destroyAllWindows()
print(hierarchy)
# ALL four terms are important t understand relationships between counters
```

```python
# RETR_TREE -> retrives all the counters nested counter,counters nested inside that and soo on
img=np.zeros((400,400,3),np.uint8)
cv2.rectangle(img,(30,150),(350,350),(200,100,155),-1)
cv2.circle(img,(220,240),50,(200,100,100),-1)
canny_edge=cv2.Canny(img,100,220)
cv2.imshow("Image",img)


contours,hierarchy=cv2.findContours(canny_edge,cv2.RETR_TREE,cv2.CHAIN_APPROX_SIMPLE,offset=(2,2))
cont_img_retr_ext=cv2.drawContours(img,contours,-1,(255,220,160),5)
cv2.imshow("Drawing of contours using Simple",cont_img_retr_ext)
cv2.imshow('Canny Edges After Contouring', canny_edge)
cv2.waitKey(0)
cv2.destroyAllWindows()
# It perfectly depicts the true natur of counters in the image
```

```python
img=np.zeros((400,400,3),np.uint8)
cv2.rectangle(img,(30,150),(350,350),(200,100,155),-1)
cv2.circle(img,(220,240),50,(200,100,100),-1)
canny_edge=cv2.Canny(img,100,220)
contours,hierarchy=cv2.findContours(canny_edge,cv2.RETR_LIST,cv2.CHAIN_APPROX_SIMPLE,offset=(2,2))

for contour in contours:
    approx=cv2.approxPolyDP(contour,0.01*cv2.arcLength(contour,True),True)
    """
    cv2.approxPolyDP(curve,epsilon,flag) -> approximates the curve with another curve 
    arg1 -> Input vector or contour line
    arg2 -> specifies the approximation accuracy
         -> If value is smaller the the approxomate shape has to be more similar to be considerd same
         -> smaller value means smaller distance between original and approximate curve
    arg3 -> if true curve is closed else not
    """
    """
    cv2.arcLength(contour,True) -> Calculates the length of the curve
    arg1 -> Inpt curve
    arg2 -> if true curve is closed else not 
    """
    corners=len(approx)

    if corners==3:
        shape_name="Triangle"
    elif corners==4:
        shape_name="Rectangle"
    elif corners==5: 
        shape_name="Pentagon"
    elif corners>5:
        shape_name="Circle"
    else:
        shape_name="Unkown"

    cv2.drawContours(img,[approx],0,(200,100,100),3)
    x=approx.ravel()[0]
    y=approx.ravel()[1] - 10
    cv2.putText(img,shape_name,(x,y),cv2.FONT_HERSHEY_COMPLEX,0.7,(255,255,245))

cv2.imshow("Contours",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
