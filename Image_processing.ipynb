#### Blurring 
- It reduces sharpness of the image and makes details of the image smoother
- Works by averaging the pixel value around each pixel
- It usess low pass filters that reduce high frequency noise and smooh fine details
- Used in noise reduction, edge smoothing etc


1. Blur 
- Take the mean of pixels within the kernel area and replaces central value with it's mean

```python
blr_img_anchor=cv2.blur(img,(3,3),(0,0))
blr_img=cv2.blur(img,(3,3))
"""
cv2.blur(image,kernel_size,anchor,bordertype)
arg1 -> image path
arg2 -> Kernel size
arg3 -> center point of the kernel , bydefault value is (-1,-1)
"""
cv2.imshow("Blurred Image anchor ",blr_img_anchor)
cv2.imshow("Blurred Image",blr_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

2. Gaussian Blur
- Works by averaging the pixel values with closer pixel with higher weight

```python
gau_blr=cv2.GaussianBlur(img,(3,5),1,2)
"""
cv2.gaussianblur(img,kernel,sigmaX,sigmaY,bordertype)
arg1 -> image 
arg2 -> width and height of the kernel ,both the values should be odd 
    It helps to determine the centre pixel , and that pixel's value will replace by the mean of values of pixels specifed in the kernel
    width blurs the image horizontally while , height blurs vertically
arg3 -> sigmaX = SD about X
arg4 -> sigmaY = SD about Y
    if arg4 is not given it is taken as same as sigmaX 
    and if 0 is passed then value will depend on size of kernel
"""
cv2.imshow("Original Image",img)
cv2.imshow("Gaussian Blur",gau_blr)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

3. Median Blur 
- Works by first sorting the pixels , then finding the median and then replaces the median value in place of center pixel of original pixels
- Helps in removing the Noise from the image


```python
noisy_img=cv2.imread(r"C:\Users\BAPS\Desktop\Python\Images\balloons_noisy.png")
med_blr=cv2.medianBlur(noisy_img,11)
"""
cv2.medianblur(img,kernel_size,bordertype)
arg1 -> Image path
arg2 -> Size of kernel , total number of pixels which will be taken to calculate the mdeian value
"""
cv2.imshow("Original Noisy Image",noisy_img)
cv2.imshow("Median Blurred",med_blr)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### Gaussian Blur vs Median Blur on noisy image

```python
noisy_gau_blr=cv2.GaussianBlur(noisy_img,(15,15),0)
cv2.imshow("Median Blurred",med_blr)
cv2.imshow("Original Noisy Image",noisy_img)
cv2.imshow("GaussianBlur on Noisy Image",noisy_gau_blr)
cv2.waitKey(0)
cv2.destroyAllWindows()

# OBSERVATION -> Median Blur removes noise very easily and blurs aggresively but gaussian blur better blurs the image in working better with fine details
```

#### Gaussian Blur vs Median Blur on normal image

```python
nrml_med_blr=cv2.medianBlur(img,15)
cv2.imshow("Median Blur",nrml_med_blr)
cv2.imshow("Original Image",img)
cv2.imshow("Gaussian Blur",gau_blr)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

4. Box Blur 

```python
box_blr=cv2.boxFilter(img,-1,(5,5),(1,1),normalize=False)
box_blr_nrm=cv2.boxFilter(img,-1,(5,5),(1,1),normalize=True)
"""
cv2.boxFilter(img,ddepth,kernel_size,anchor,normalize_flag,bordertype)
arg1-> img
arg2-> depth of dest img (-1 to use src.depth() i.e. depth of source img)
arg3-> size of kernel 
arg5 -> if true kernel is normalized,else not
"""
cv2.imshow("Box Blurred Image normalize",box_blr_nrm)
cv2.imshow("Box Blurred Image without normalize",box_blr)
cv2.imshow("Original Image",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

##### Filter2D() -> we can blur,sharp and detect edge on the image

```python
sharpening_kernel=np.array([
    [0,-1,0],
    [-1,5,-1],
    [0,-1,0]
])
sharped_img=cv2.filter2D(med_blr,-1,sharpening_kernel)
"""
cv2.filter2D(img,ddepth,kernel)
arg3 -> np array which has tells the kernel how the central pixel should be treated
"""
cv2.imshow("Sharped Image",sharped_img)
cv2.imshow("Original",med_blr)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

##### Types of Kernel

```python
# Identity Kernel -> Keeps the image as it is,but by changing the values of central pixel brightness of the image can changed
identity_kernel=np.array([
    [0,0,0],
    [0,1,0],
    [0,0,0]
])
identity_img=cv2.filter2D(img,-1,identity_kernel)
cv2.imshow("Identity Image",identity_img)
cv2.imshow("Original",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Edge Detection -> Extracts the edge from the image,it can be used to enchance the foreground
edge_kernel=np.array([
    [-1,-1,-1],
    [-1,8,-1],
    [-1,-1,-1]
])
edge_img=cv2.filter2D(img,-1,edge_kernel)
cv2.imshow("edge Image",edge_img)
cv2.imshow("Original",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Sharpen Kernel -> This enhances the central pixel and is used to sharpen the image
sharpening_kernel=np.array([
    [0,-1,0],
    [-1,5,-1],
    [0,-1,0]
])
sharped_img=cv2.filter2D(med_blr,-1,sharpening_kernel)
cv2.imshow("Sharped Image",sharped_img)
cv2.imshow("Original",med_blr)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Box Blur -> Box blurs the image,changing the denominator changes the brightness along with the boxblur extent
# for blur without changing anything else denominatr should be square of size of array shape
boxblur_kernel=np.ones((5,5))/25
boxblur_kernel_img=cv2.filter2D(img,-1,boxblur_kernel)
cv2.imshow("BoxBlurred Image",boxblur_kernel_img)
cv2.imshow("Original",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
random_kernel=np.random.randint(2,size=(3,3))
random_kernel_img=cv2.filter2D(img,-1,random_kernel)
cv2.imshow("Random Operation",random_kernel_img)
cv2.imshow("Original",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Gaussian blur Kernel 
gaussian_kernel=np.array([
    [1,4,6,4,1],
    [4,16,24,16,4],
    [6,24,36,24,6],
    [4,16,24,16,4],
    [1,4,6,4,1]
])/256

gaussian_kernel_img=cv2.filter2D(img,-1,gaussian_kernel)
cv2.imshow("Gaussian kernel blur",gaussian_kernel_img)
cv2.imshow("Original",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cv2.imshow("Gaussian kernel blur",gaussian_kernel_img)
cv2.imshow("Gaussian Blur",gau_blr)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Bilateral Filtering -> smooths the image,reduces noise while preserving the noise
bil_fil=cv2.bilateralFilter(noisy_img,200,200,500)
"""
cv2.bilateralFilter(img,diameter,sigmaColor,sigmaSpace)
arg1 -> img
arg2 -> diameter of each pixel neighbourhood
arg3 -> sigma value in color space,greater the value colors farther to each other will start to mix
arg4 -> sigma value in coordinate space, greater the value further pixels will start to mix
"""
cv2.imshow("Bilateral filtering",bil_fil)
cv2.imshow("Median Blur",med_blr)
cv2.imshow("Original",noisy_img)
cv2.waitKey(0)
cv2.destroyAllWindows()

# OBSERVATION -> while this removes the noise but it is not as effective as median blur
```

#### Thresholding 
- Used to segment objects from background
- If value below threshold 0(black) if above 255(white)

```python
ret,thres_img=cv2.threshold(img,150,255,cv2.THRESH_BINARY)
# To be use with black and white images
"""
flag,img=cv2.threshold(img,threshold,max value,threshold_functions)
arg1 -> image
arg2 -> Threshold value used to classify pixels
arg3 -> value given to pixels above the threshold
arg4 -> Which function will be used for thresholding
"""
# ret -> stores the optimal value for thresholding 
cv2.imshow("Original Image",img)
cv2.imshow("Black and White Image",bw_img)
cv2.imshow("Binary Threshold Image",thres_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
# OBSERVATION -> Can be used with color images but it increases the saturation of the image 
```

##### Different Threholdig techniques

```python
# THRESH_BINARY -> Value above threshold 255 , below threshold 0
ret,thres_img_tb=cv2.threshold(bw_img,150,255,cv2.THRESH_BINARY)
cv2.imshow("Black and White Image",bw_img)
cv2.imshow("Binary Threshold Image",thres_img_tb)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# THRESH_BINARY_INV -> Value above threshold 0 , below threshold 255
ret,thres_img_tbi=cv2.threshold(bw_img,150,255,cv2.THRESH_BINARY_INV)
cv2.imshow("Black and White Image",bw_img)
cv2.imshow("Inverse Binary Threshold Image",thres_img_tbi)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# THRESH_TRUNC -> Above threshold are set to value of threshold ,below remains unchanged
ret,thres_img_tt=cv2.threshold(bw_img,100,255,cv2.THRESH_TRUNC)
cv2.imshow("Black and White Image",bw_img)
cv2.imshow("Truncate Threshold Image",thres_img_tt)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# THRESH_TOZERO -> Below threshold set to 0, above remains unchanged
ret,thres_img_ttz=cv2.threshold(bw_img,150,255,cv2.THRESH_TOZERO)
cv2.imshow("Black and White Image",bw_img)
cv2.imshow("Tozero Threshold Image",thres_img_ttz)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# THRESH_TOZERO_INV -> above threshold set to 0, below remians unchanged
ret,thres_img_ttzi=cv2.threshold(bw_img,150,255,cv2.THRESH_TOZERO_INV)
cv2.imshow("Black and White Image",bw_img)
cv2.imshow("Tozero Threshold Image",thres_img_ttzi)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# THRESH_TRIANGLE -> Uses triangle method to find the optimal threshold method
ret,thres_img_ttri=cv2.threshold(bw_img,100,255,cv2.THRESH_TRIANGLE)
cv2.imshow("Black and White Image",bw_img)
cv2.imshow("Triangle Threshold Image",thres_img_ttri)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# THRESH_OTSU -> Used when image has two distinct pixel groups , it calculates the optimal thresholding by analyzing the image histogram
ret,thres_img_to=cv2.threshold(bw_img,0,255,cv2.THRESH_BINARY + cv2.THRESH_OTSU)
"""
Threshold is skipped for this
Any thresholding technique will workwith otsu
"""
cv2.imshow("Black and White Image",bw_img)
cv2.imshow("OTSU Threshold Image",thres_img_to)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
ret,thres_img_tm=cv2.threshold(bw_img,250,200,cv2.THRESH_BINARY | cv2.THRESH_MASK)
cv2.imshow("Black and White Image",bw_img)
cv2.imshow("Mask Threshold Image",thres_img_tm)
cv2.waitKey(0)
cv2.destroyAllWindows()
print(ret,thres_img_tm,"\n",bw_img)
```

##### Adaptive Thresholding 
- Divids the image into small regions and thresholding is applied for each region 
- Segments image effectively when lighting changes are there in the image

```python
# ADAPTIVE_THRESH_MEAN_C -> Threshold for each pixel is calculated as mean of surrounding blocks - constant 
bw_img1=cv2.imread(r"C:\Users\BAPS\Desktop\Python\Images\Halfshade.jpg",cv2.IMREAD_GRAYSCALE)
thres_img_mean=cv2.adaptiveThreshold(bw_img1,255,cv2.ADAPTIVE_THRESH_MEAN_C,cv2.THRESH_BINARY,21,5)
"""cv2.adaptiveThreshold(img,max_value,adaptive_threshold_method,threshold_function,block_size,constant)
arg1 -> image
arg2 -> max value given to pixel which satisfies the threshold
arg3 -> which adaptive method will be used
arg4 -> Which thresholding function will be used(cv2.THRESH_BINARY,cv2.THRESH_BINARY_INV)
arg5 -> size of surrounding blocks(must be odd)
arg6 -> value of constant
"""
cv2.imshow("Black and White Image",bw_img1)
cv2.imshow("Adaptive Mean Threshold Image",thres_img_mean)
cv2.waitKey(0)
cv2.destroyAllWindows()
# OBSERVATION -> increasing the value of constant it removes background outlines
#             -> increasing the block_size increases the background details
```

```python
# ADAPTIVE_THRESH_GAUSSIAN_C -> Threshold for each pixel is weighted sum of surrounding pixels - constant
thres_img_gaussian=cv2.adaptiveThreshold(bw_img1,255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,21,5)
cv2.imshow("Black and White Image",bw_img1)
cv2.imshow("Gaussian Mean Threshold Image",thres_img_gaussian)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cv2.imshow("Black and White Image",bw_img1)
cv2.imshow("Gaussian Mean Threshold Image",thres_img_gaussian)
cv2.imshow("Adaptive Mean Threshold Image",thres_img_mean)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### BitWise Operations on Image
- Are used for image manipulation and extracting essential parts in the image


```python
img1=np.zeros((400,400,3),np.uint8)
img2=np.zeros((400,400,3),np.uint8)

cv2.circle(img1,(200,200),100,(255,100,20),-1)
cv2.rectangle(img2,(30,150),(350,350),(255,100,20),-1)
cv2.imshow("Cricle",img1)
cv2.imshow("Rectangle",img2)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# AND -> extracts the intersecting part of the image
# mask = np.zeros((400,400), dtype="uint8")
and_img=cv2.bitwise_and(img1,img2,mask=None)
"""
cv2.bitwise_and(img1,img2,mask)
arg1 -> image 1
arg2 -> image 2
arg3 -> mask = user defined array to hide particular part of image
"""
cv2.imshow("Bitwise And",and_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
# OBSERVATION -> if this is used to colors of both the images also gets AND gated
#             -> Shape of both the images should be same.
```

```python
# OR -> combines both the images
or_img=cv2.bitwise_or(img1,img2,mask=None)
cv2.imshow("Bitwise OR",or_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# XOR -> IF same 0, different is 1
xor_img=cv2.bitwise_xor(img1,img2,mask=None)
cv2.imshow("Bitwise XOR",xor_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# NOT -> negation
not_img=cv2.bitwise_not(img2,mask=None)
cv2.imshow("Bitwise NOT",not_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
# OBSERVATION -> If we pass two images in arguments the Not of first image will get stored in the second image
```
