#### This file contains face detection

```python
import cv2
import numpy as np
```

```python
face_cascade=cv2.CascadeClassifier(r"C:\Users\BAPS\Desktop\Python\Machine_Learning\OpenCV\haarcascade_frontalface_default.xml")
"""
cv2.CascadeClassifier(model_path) -> loads the model 
"""
cap=cv2.VideoCapture(0)

while True:
    ret,frame=cap.read()
    if not ret:
        print("OOPS!!!!!!!")
        break
    bw_frame=cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)

    faces=face_cascade.detectMultiScale(bw_frame,scaleFactor=1.1,minNeighbors=15)
    """
    model.detectMultiScale(img,scaling_factor,minimun_neighbours,minvalue,maxvalue)
    arg1 -> image
    arg2 -> factor by which the frames would get zoomed , until object is not detected(like the area would be reduced by this amount)
    arg3 -> minimun number of neighbouring rectangle that must overlap with candidate rectangle to be considered an valid rectangle 
    arg4 -> Minimum object value,objects below this will be ignored
    arg5 -> Maximum object value,objects above this will be igmored
    """

    for (x,y,w,h) in faces:
        cv2.rectangle(frame, (x,y),(x+w+20,y+h+20),(0,0,255),3)
    
    cv2.imshow("Face Detection",frame)

    if cv2.waitKey(1) & 0XFF==ord("q"):
        print("Quitting")
        break

cap.release()
cv2.destroyAllWindows()
```

### Template Matching
- Using this we can create bounding boxes on objects based on similarity and we have to use any model

```python
img = cv2.imread(r"Practice.jpg",0)
img=cv2.resize(img,None,fx=0.7,fy=0.7,interpolation=cv2.INTER_LINEAR)
cv2.imshow("Image",img)

template_img=cv2.resize(cv2.imread(r"cropped_image_roi.png",0),None,fx=0.7,fy=0.7,interpolation=cv2.INTER_LINEAR)
h,w=template_img.shape
methods=[cv2.TM_CCOEFF,cv2.TM_CCOEFF_NORMED,cv2.TM_CCORR,cv2.TM_CCORR_NORMED,cv2.TM_SQDIFF,cv2.TM_SQDIFF_NORMED]
# OBSERVATION -> The values of these constants are internally defined by OpenCv as -> [4, 5, 2, 3, 0, 1] respectively
"""
cv2.TM_CCOEFF[4] -> Corelation Coeficient = removes the average intensity i.e. ignores overall brightness differencess (big values= better match)
cv2.TM_CCOEFF_NORMED[5] -> normalized version of CCOEFF
cv2.TM_CCORR[2] -> Cross corelation = direct multiplication of pixels (big values= better match)
cv2.TM_CCORR_NORMED[3] -> Normalised version of CCORR
cv2.TM_SQDIFF[0] -> Square difference = compares pixel differences (small values= better match)
cv2.TM_SQDIFF_NORMED[1] -> Normalised version of SQDIFF
"""
for method in methods:
    img2=img.copy()
    result= cv2.matchTemplate(img,template_img,method)
    """
    cv2.matchTemplate(image,template_image,method,mask) -> matches the template by sliding the template over the image ,and stores the result in a matrix 
    arg1 -> image from which template would be checked
    arg2 -> template image
    arg3 -> comparision method
    """
    min_val,max_val,min_loc,max_loc=cv2.minMaxLoc(result)
    """
    cv2.minMaxLoc(image) -> finds the min and max element values and their positions
    arg - source image 
    """
    # """
    # returns
    # min_val -> pointer to minimun value
    # max_val -> pointer to maxmimum value
    # min_loc -> pointer to minimum value location (2D)
    # max_loc -> pointer to maximum value location (2D)
    # """
    if method in [cv2.TM_SQDIFF,cv2.TM_SQDIFF_NORMED]:
        location = min_loc
    else:
        location=max_loc
    bottom_right=(location[0]+w,location[1]+h)
    string=f"Match {methods[method]}"
    cv2.rectangle(img,location,bottom_right,255,5)
    cv2.imshow(string,img2)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
# OBSERVATION -> There is no for sure a better method , each time you have to check all methods , and the one that better results move ahead with that method
```
